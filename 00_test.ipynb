```python
# student grade A-above95 marks, B-85 to 95 marks, C-75 to 85 marks, D-65 to 75 marks, E-55 to 65 marks, F-below 55 marks
# if elif else format
marks = float(input("Enter your marks: "))
if marks > 95:
    grade = "A"
elif marks >= 85:
    grade = "B"
elif marks >= 75:
    grade = "C"
elif marks >= 65:
    grade = "D"
elif marks >= 55:
    grade = "E"
else:
    grade = "F"
print(f"Your grade is: {grade}")
```
