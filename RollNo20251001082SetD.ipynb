```python
import math

raw_kc = input("Enter Kc: ")
print("Raw input type:", type(raw_kc))

try:
    kc = float(raw_kc)
except ValueError:
    print("Error: Kc must be a number.")
else:
    print("Converted value type:", type(kc))

    if not math.isfinite(kc) or kc <= 0:
        print("Error: Kc must be a finite number greater than 0.")
    else:
        # Kc*x^2 - (2.5*Kc + 1)*x + 1.5*Kc = 0
        a = kc
        b = -(2.5 * kc + 1)
        c = 1.5 * kc

        discriminant = b**2 - 4 * a * c
        print(f"a = {a}")
        print(f"b = {b}")
        print(f"c = {c}")
        print(f"Discriminant = {discriminant}")

        sqrt_discriminant = math.sqrt(discriminant)
        root1 = (-b + sqrt_discriminant) / (2 * a)
        root2 = (-b - sqrt_discriminant) / (2 * a)
        print(f"Root 1 = {root1}")
        print(f"Root 2 = {root2}")

        # B is limiting, so the physically meaningful x must be in [0, 1].
        valid_roots = [root for root in (root1, root2) if 0 <= root <= 1]

        if not valid_roots:
            print("Error: No physically meaningful root was found.")
        else:
            x = valid_roots[0]

            equilibrium_a = 1.5 - x
            equilibrium_b = 1.0 - x
            equilibrium_c = x
            conversion_b = (x / 1.0) * 100

            print(f"\nEquilibrium [A] = {equilibrium_a:.3f} mol/L")
            print(f"Equilibrium [B] = {equilibrium_b:.3f} mol/L")
            print(f"Equilibrium [C] = {equilibrium_c:.3f} mol/L")
            print(f"Conversion of B = {conversion_b:.3f}%")

            if conversion_b >= 70:
                classification = "High"
            elif conversion_b >= 40:
                classification = "Moderate"
            else:
                classification = "Low"

            print("Conversion classification:", classification)
```

```python
def factorial(n):
    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("n must be an integer.")
    if n < 0:
        raise ValueError("Factorial is not defined for negative values.")
    if n == 0 or n == 1:  # 0! = 1
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("n must be an integer.")
    if not isinstance(r, int) or isinstance(r, bool):
        raise TypeError("r must be an integer.")
    if n < 0 or r < 0 or r > n:
        raise ValueError("Invalid input: require n >= 0 and 0 <= r <= n.")

    return factorial(n) // (factorial(r) * factorial(n - r))


for n, r in [(9, 3), (6, 8)]:
    try:
        print(f"n = {n}, r = {r}: {combination(n, r)} possible mixtures")
    except (TypeError, ValueError) as error:
        print(f"n = {n}, r = {r}: {error}")
```

```python
def fib_iter(n):
    if n < 0:
        raise ValueError("n must be non-negative.")

    sequence = []
    first, second = 0, 1

    for _ in range(n):
        sequence.append(first)
        first, second = second, first + second

    return sequence


def fib_rec(k):
    if k < 0:
        raise ValueError("k must be non-negative.")
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


n = 10
iterative_values = fib_iter(n)
recursive_values = [fib_rec(k) for k in range(n)]

if iterative_values == recursive_values:
    print("Both functions agree.")
else:
    print("The functions do not agree.")

print(f"\n{'Sample No.':<12} | {'Fibonacci tag':<13} | {'Sample time (min)'}")
print("-" * 50)

for sample_no, fib_value in enumerate(iterative_values, start=1):
    sample_time = fib_value * 4
    print(f"{sample_no:<12} | {fib_value:<13} | {sample_time}")
```
